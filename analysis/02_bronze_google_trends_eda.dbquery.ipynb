```sql
SELECT * 
FROM info_env_jordan.bronze.google_trends
LIMIT 100 ; 

SELECT search_term,
    MIN(trend_score) min_score,
    MAX(trend_score) max_score,
    MIN(trend_value_display) min_value,
    MAX(trend_value_display) max_value
FROM info_env_jordan.bronze.google_trends
GROUP BY search_term
ORDER BY 1 ASC    


SELECT search_term,
    MIN(trend_score) min_score,
    MAX(trend_score) max_score,
    MIN(trend_value_display) min_value,
    MAX(trend_value_display) max_value
FROM info_env_jordan.bronze.google_trends
WHERE (trend_score > 0 OR trend_value_display > 0)
GROUP BY search_term
ORDER BY 1 ASC    


-- trend_score is always the same as trend_value_display
SELECT COUNT(*)
FROM info_env_jordan.bronze.google_trends
WHERE (trend_score <> trend_value_display)


-- Are there any trends?
    -- we are seeing a spike in 2025 and even greater spike in 2026
SELECT search_term,
    YEAR(event_Date) AS year,
    AVG(trend_score) avg_score
FROM info_env_jordan.bronze.google_trends
GROUP BY search_term,
    YEAR(event_Date)
ORDER BY 1 ASC, 2 ASC  


-- most important topic in 2026 is unemployment 
SELECT search_term,
    YEAR(event_Date) AS year,
    AVG(trend_score) avg_score
FROM info_env_jordan.bronze.google_trends
WHERE YEAR(event_Date) = 2026
GROUP BY search_term,
    YEAR(event_Date)
ORDER BY 3 DESC     
```
